```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
iris = pd.read_csv("./iris.csv")
iris
```

```python
header = iris.columns.values.tolist()
header[0:4]
```

```python
# min-max scaling
for i in header[0:4]:
    iris[i] = (iris[i].values - iris[i].values.min()) / (iris[i].values.max() - iris[i].values.min())
```

```python
iris
```

```python
iris_setosa = iris.iloc[0:50, 0:5]
iris_virginica = iris.iloc[50:100, 0:5]
iris_versicolor = iris.iloc[100:150, 0:5]
```

```python
# setosa = 0, virginica = 1, versicolor = 2
iris_setosa["class"] = 0
iris_virginica["class"] = 1
iris_versicolor["class"] = 2
```

```python
iris_setosa.iloc[0:50, 0:4].plot()
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.title("Setosa")
```

```python
iris_versicolor.iloc[0:50, 0:4].plot()
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.title("Versicolor")
```

```python
iris_virginica.iloc[0:50, 0:4].plot()
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.title("Virginica")
```

```python
iris_scaled = pd.concat([iris_setosa, iris_virginica, iris_versicolor])
```

```python
iris_scaled.to_csv("./iris_scaled.csv", index=False)
```
